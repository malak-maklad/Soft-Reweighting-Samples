```python
columns_to_inspect = ["DS", "Tendency", "DP", "Nzeros", "Nmax"]

for col in columns_to_inspect:
    print(f"\n{col} Value Counts:")
    print(X[col].value_counts().sort_index())
```

```python
duplicate_mask = X.duplicated(keep=False)

duplicate_records = full_data.loc[duplicate_mask]

display(duplicate_records.sort_values(["NSP", "CLASS"]))
```

```python
duplicate_group_summary = full_data.groupby(
    list(X.columns),
    dropna=False
).agg(
    record_count=("NSP", "size"),
    nsp_unique=("NSP", "nunique"),
    class_unique=("CLASS", "nunique"),
    nsp_values=("NSP", lambda s: sorted(s.unique())),
    class_values=("CLASS", lambda s: sorted(s.unique()))
)

duplicate_group_summary = duplicate_group_summary[
    duplicate_group_summary["record_count"] > 1
]

display(duplicate_group_summary)
```

```python
nsp_conflicts = duplicate_group_summary[
    duplicate_group_summary["nsp_unique"] > 1
]

print("Number of groups with conflicting NSP:", len(nsp_conflicts))
display(nsp_conflicts)
```

```python
analysis_data = X.copy()
analysis_data["NSP"] = y["NSP"]

features_to_compare = [
    "LB", "ASTV", "MSTV",
    "ALTV", "MLTV", "Variance"
]

target_summary = analysis_data.groupby("NSP")[
    features_to_compare
].agg(["mean", "median", "std"])

display(target_summary.T)
```

```python
correlation_matrix = X.corr(method="pearson")

display(correlation_matrix.round(2))
```

```python
upper_triangle = correlation_matrix.where(
    np.triu(
        np.ones(correlation_matrix.shape),
        k=1
    ).astype(bool)
)

correlation_pairs = upper_triangle.stack().reset_index()

correlation_pairs.columns = [
    "Feature_1",
    "Feature_2",
    "Correlation"
]

correlation_pairs["Absolute_Correlation"] = (
    correlation_pairs["Correlation"].abs()
)

correlation_pairs = correlation_pairs.sort_values(
    by="Absolute_Correlation",
    ascending=False
)

display(correlation_pairs.head(15).round(3))
```

```python
spearman_correlation = X.corr(method="spearman")

upper_triangle_spearman = spearman_correlation.where(
    np.triu(
        np.ones(spearman_correlation.shape),
        k=1
    ).astype(bool)
)

spearman_pairs = upper_triangle_spearman.stack().reset_index()

spearman_pairs.columns = [
    "Feature_1",
    "Feature_2",
    "Correlation"
]

spearman_pairs["Absolute_Correlation"] = (
    spearman_pairs["Correlation"].abs()
)

spearman_pairs = spearman_pairs.sort_values(
    by="Absolute_Correlation",
    ascending=False
)

display(spearman_pairs.head(15).round(3))
```

# Exploratory Data Analysis (EDA)
## Cardiotocography (CTG) Dataset

### 1. Dataset Overview

The Cardiotocography (CTG) dataset was obtained from the UCI Machine Learning Repository.

The dataset contains 2,126 records and 21 input features, along with two target variables: `CLASS` and `NSP`.

For this project, `NSP` is selected as the primary target variable, while `CLASS` is excluded from the input features to avoid using another target variable as a predictor.

**Dataset dimensions:**
- Number of records: 2,126
- Number of input features: 21
- Number of target variables: 2
- Primary target: `NSP`

### 2. Feature Description

The input features describe different characteristics of fetal heart rate and uterine contraction measurements.

They can be grouped into the following categories:

| Feature Group | Features |
|---|---|
| Baseline Heart Rate | `LB` |
| Accelerations | `AC` |
| Fetal Movements | `FM` |
| Uterine Contractions | `UC` |
| Decelerations | `DL`, `DS`, `DP` |
| FHR Variability | `ASTV`, `MSTV`, `ALTV`, `MLTV` |
| Histogram Characteristics | `Width`, `Min`, `Max`, `Nmax`, `Nzeros`, `Mode`, `Mean`, `Median`, `Variance`, `Tendency` |

### 3. Data Types and Descriptive Statistics

The dataset contains:
- 8 features with `float64` data type.
- 13 features with `int64` data type.
- All 21 input features contain 2,126 non-null observations.

Descriptive statistics were examined using the `describe()` function to understand the distributions, central tendencies, and ranges of the features.

**Key observations:**

- Features have considerably different numerical scales.
- Some features, such as `FM`, `DS`, and `DP`, contain many zero values.
- Features such as `MSTV`, `ALTV`, `MLTV`, and `Variance` show relatively wide ranges.
- `Tendency` takes values from -1 to 1.

The differences in feature scales should be considered during the feature preparation and modeling stages.

### 4. Missing Values Analysis

Missing values were checked in both input features and target variables.

**Results:**
- Missing values in input features: 0
- Missing values in `CLASS`: 0
- Missing values in `NSP`: 0

Therefore, no missing values were detected in the original dataset.

### 5. Duplicate Records Analysis

Duplicate observations were examined at two levels:

- Duplicate rows in input features (`X`): 15
- Duplicate complete records (features and targets): 11

Further investigation identified 13 groups of repeated feature combinations. Some repeated feature combinations had conflicting target labels.

In particular, two groups contained identical input features but different `NSP` labels.

**Observation:**

These cases may introduce ambiguity during model training because identical feature values are associated with different target labels.

No duplicate records were removed during EDA. Their handling should be considered during the data preparation stage.

### 6. Unique Values and Low-Cardinality Features

The number of unique values was examined for all input features.

Some features have relatively few unique values:

- `DS`: 2 unique values.
- `Tendency`: 3 unique values.
- `DP`: 6 unique values.
- `Nzeros`: 9 unique values.
- `Nmax`: 18 unique values.

The value distributions of selected features were also examined.

**Key observations:**

- `DS` contains 2,119 zero values out of 2,126 observations.
- `DP` contains 1,948 zero values.
- `Nzeros` contains 1,624 zero values.
- `Tendency` takes three values: -1, 0, and 1.

These findings indicate that some features are highly concentrated around zero or have a limited number of distinct values.

Zero values were not treated as missing values because they may represent valid measurements.

### 7. Feature Distribution Analysis

### 8. Target Variable Analysis

The `NSP` variable was selected as the primary prediction target.

Its classes are:

- 1: Normal
- 2: Suspect
- 3: Pathologic

#### 8.1 Target Distribution

| Class | Description | Count | Percentage |
|---|---|---:|---:|
| 1 | Normal | 1,655 | 77.85% |
| 2 | Suspect | 295 | 13.88% |
| 3 | Pathologic | 176 | 8.28% |
| **Total** | | **2,126** | **100%** |

#### 8.2 Class Imbalance

The target distribution shows a noticeable class imbalance.

The Normal class represents approximately 77.85% of the observations, while the Pathologic class represents only 8.28%.

The Normal class has approximately 9.4 times as many observations as the Pathologic class.

**Potential impact:**

A model may become biased toward predicting the majority class. Therefore, accuracy alone may not provide a reliable evaluation of model performance.

Metrics such as Macro F1-score, Balanced Accuracy, and per-class Recall should be considered during model evaluation, especially for the minority Pathologic class.

No resampling or class balancing techniques were applied during EDA.

### 9. Feature vs. Target Analysis

Selected numerical features were compared across the three `NSP` classes using mean, median, and standard deviation.

The following observations were identified:

- `LB`: The Suspect class has a higher mean and median than the other two classes.
- `ASTV`: Mean and median values increase from Normal to Suspect to Pathologic.
- `MSTV`: The Suspect class has lower values, while the Pathologic class has higher values.
- `ALTV`: The Suspect class has the highest mean, while the Pathologic class shows a larger standard deviation.
- `MLTV`: The Pathologic class has noticeably lower mean and median values.
- `Variance`: The Pathologic class has substantially higher mean and median values than the other classes.

These differences suggest that selected features may have different distributions across the target classes.

**Note:** These are descriptive observations only. They do not establish statistical significance or prove that a feature is important for prediction.

### 11. Overall EDA Summary

The exploratory analysis provided the following main findings:

1. The dataset contains 2,126 observations and 21 input features.
2. No missing values were detected.
3. Duplicate feature combinations exist, including cases with conflicting target labels.
4. Several features contain a high proportion of zero values.
5. Some numerical features exhibit skewed or concentrated distributions.
6. The target variable `NSP` is imbalanced, with Normal as the dominant class.
7. Selected features show different descriptive patterns across target classes.
8. Strong correlations exist among several input features, indicating potential redundancy.

### 12. EDA Limitations and Recommendations

Based on the observations, the following points should be considered during subsequent preparation and modeling stages:

- Investigate repeated feature combinations with conflicting target labels before deciding how to handle duplicates.
- Preserve valid zero values unless their meaning is verified to indicate missing or invalid measurements.
- Consider the target class imbalance when selecting evaluation metrics and designing experiments.
- Examine potential feature redundancy during feature preparation and model evaluation.
- Consider feature scaling where required by the selected algorithms.
- Avoid removing features solely based on correlation values.

**Final Note:** No data cleaning, feature transformation, feature selection, or resampling operations were performed during this EDA stage. These tasks are reserved for the subsequent data preparation stages.
