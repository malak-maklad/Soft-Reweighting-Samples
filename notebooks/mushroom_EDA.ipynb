```python
print("Features to exclude from model training:")
for col in constant_features:
    print(f"- {col}: {df_clean[col].unique()}")
```

### Data Cleaning Summary

The dataset was inspected to identify missing values, duplicate records, constant features, and invalid categorical values.

The following observations were identified:

1. **Missing Values:** The `stalk-root` attribute contains 2,480 missing values represented by `?`, accounting for approximately 30.53% of the dataset.
2. **Missingness Analysis:** Missing values in `stalk-root` are associated with a different target distribution compared with non-missing observations.
3. **Duplicate Records:** No duplicate rows were found.
4. **Constant Features:** The `veil-type` attribute contains only one unique value (`p`) across all observations.
5. **Invalid Categories:** No unexpected categorical values were identified during the inspection.

**Cleaning Decisions:**

- Retain all observations, including those with missing `stalk-root` values.
- Preserve `?` as a separate categorical value to maintain compatibility with the original AUMtree preprocessing implementation.
- Retain `veil-type` in the original dataset but exclude it from model input features due to zero variance.
- No rows or columns have been modified or deleted at this stage.

The resulting dataset retains its original shape of 8,124 rows and 23 columns.

### Feature and Target Definition

- **Input Features (X):** The dataset contains 22 categorical features.
- **Target Variable (y):** The `class` column represents the mushroom class:
  - `e`: Edible
  - `p`: Poisonous
- **Feature Type:** All input features are nominal categorical variables.
- **Prediction Task:** Binary Classification.

```python
X = df_clean.drop(columns=["class"])
y = df_clean["class"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
print(y.value_counts())
print("\nTarget percentages:")
print(y.value_counts(normalize=True) * 100)
```

```python

X_encoded = pd.get_dummies(
    X,
    drop_first=False,
    dtype=int
)


y_encoded = y.map({
    "e": 0,
    "p": 1
}).astype(int)

print("Original X shape:", X.shape)
print("Encoded X shape:", X_encoded.shape)

print("\nEncoded target distribution:")
print(y_encoded.value_counts())

print("\nMissing values in encoded features:", X_encoded.isnull().sum().sum())
print("Missing values in encoded target:", y_encoded.isnull().sum())

print("\nFeature data types:")
print(X_encoded.dtypes.value_counts())
```

```python

print("Target mapping:")
print(y.value_counts().sort_index())

print("\nEncoded target mapping:")
print(y_encoded.value_counts().sort_index())


constant_columns = X_encoded.columns[
    X_encoded.nunique() == 1
]

print("\nConstant encoded columns:")
print(constant_columns.tolist())

print("\nNumber of constant columns:", len(constant_columns))
```

```python
print("Number of features:", X_encoded.shape[1])

print("\nUnique values in encoded features:")
print(np.unique(X_encoded.values))

print("\nAre all features binary?")
print(X_encoded.isin([0, 1]).all().all())

print("\nAre X and y lengths equal?")
print(len(X_encoded) == len(y_encoded))
```

```python
X_final = X_encoded.to_numpy(dtype=np.int32)
y_final = y_encoded.to_numpy(dtype=np.int32)


print("Final X shape:", X_final.shape)
print("Final y shape:", y_final.shape)

print("X dtype:", X_final.dtype)
print("y dtype:", y_final.dtype)

print("Target classes:", np.unique(y_final))
```

```python
preprocessed_df = X_encoded.copy()
preprocessed_df["label"] = y_encoded

preprocessed_df.to_csv(
    "mushroom_preprocessed.csv",
    index=False
)

print("Dataset saved successfully!")
print("Saved shape:", preprocessed_df.shape)
```
